```python
import json
from pathlib import Path
from IPython.display import display, Image
report=json.loads(Path("outputs_v2/metrics.json").read_text())
print("Selected model:", report["selected_model"])
print("Validation threshold:", report["selected_threshold_from_validation"])
print("Test ROC-AUC:", round(report["test"]["roc_auc"], 4))
print("Test AP:", round(report["test"]["average_precision"], 4))
print("Cost at 0.50:", report["test"]["threshold_0_5"]["cost"])
print("Cost at selected threshold:", report["test"]["selected_threshold"]["cost"])
print("SHAP:", report.get("shap_status"))
for name in ["confusion_matrix.png", "roc_curve.png", "precision_recall_curve.png", "validation_cost_curve.png", "shap_summary.png"]:
    path=Path("outputs_v2")/name
    if path.exists():
        print(name)
        display(Image(filename=str(path)))
```
